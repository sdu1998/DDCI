```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
c = 0.2
Bs = [5, 30]

# Fixed size
figsize = (6,4)

for B in Bs:
    x = np.linspace(-1.8*B, 1.8*B, 800)
    y1 = (np.abs(x)/(2*B)) * np.tanh(c * x)
    y2 = (1 - np.abs(x)/(2*B)) * np.tanh(c * x)
    
    plt.figure(figsize=figsize)
    plt.plot(x, y1, label=r'$|x|/2B \cdot \tanh(c x)$')
    plt.plot(x, y2, label=r'$(1 - |x|/2B)\cdot \tanh(c x)$')
    plt.axhline(0, linewidth=0.8, color='black')
    plt.axvline(0, linewidth=0.8, color='black')
    plt.title(fr'c = {c}, B = {B}')
    plt.xlabel('x')
    plt.ylabel('value')
    plt.legend()
    plt.grid(True, linewidth=0.3)
    plt.show()
```
